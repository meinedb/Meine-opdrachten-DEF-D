```python
# verwerk hier jullie kalibratiemetingen

afstand = np.array([10,20,30,40,50,60])
data = np.array([270,590,910,1190,1470,1760])

def function(x, a, b):
    return a*x + b
                   
values, covariance = curve_fit(function,data,afstand)
x_test = np.linspace(min(data),max(data),1000)
y_fit= function(x_test,*values)
plt.figure()
plt.plot(data,afstand,'b.')
plt.plot(x_test,y_fit,'r-')
plt.show()

print(values[0],values[1])
```
